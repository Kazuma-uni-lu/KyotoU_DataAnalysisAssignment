Let's extract the relevant columns:

```python

df = df0[  ['Rented Bike Count','Hour','Temperature(?C)','Humidity(%)','Wind speed (m/s)','Visibility (10m)','Dew point temperature(?C)','Solar Radiation (MJ/m2)','Rainfall(mm)','Snowfall (cm)',
            'Seasons','Holidays','Functioning Day']  ]

df.describe()
```

<br>
<br>

Next let's use the `rename` function to give the columns simpler variable names:

```python
df = df.rename( columns={'Rented Bike Count':'Bike_Count', 'Temperature(?C)':'Temperature','Humidity(%)':'Humidity','Wind speed (m/s)':'Wind Speed','Visibility (10m)':'Visibility','Dew point temperature(?C)':'Dew Point Temperature','Solar Radiation (MJ/m2)':'Solar Radiation','Rainfall(mm)':'Rainfall','Snowfall (cm)':'Snowfall'} )

df.describe()
```
